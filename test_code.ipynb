```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv('sample_data.csv')
```

```python
#Validate the fiels used
assert df["Product Category"].notna().all()
assert pd.api.types.is_numeric_dtype(df['Total Amount'])


catergory_sales= (
    df.groupby('Product Category')['Total Amount']
        .sum()
        .sort_values(ascending=False)
)


plt.figure(figsize=(10,6))
catergory_sales.plot(kind= 'bar', color= 'steelblue')
plt.title('Total sales by Product Catergory')
plt.xlabel('Product Catergory')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
